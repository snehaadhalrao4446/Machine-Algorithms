#creating a ML model to predict if student will get job or not

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

step 1 import file

```python
df=pd.read_csv('/content/placement.csv')
df
```

step 2 plot a graph

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package in lpa')
```

step 3 seprate X and Y

```python
x=df.iloc[:,0:1]
x
```

```python
y=df.iloc[:,-1]
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)

from sklearn.linear_model import LinearRegression
lr = LinearRegression()

lr.fit(X_train,y_train)

X_test
y_test

lr.predict(X_test.iloc[0].values.reshape(1,1))

from sklearn.metrics import mean_squared_error, r2_score,mean_absolute_error

y_pred = lr.predict(X_test)

print("MSE:", mean_squared_error(y_test, y_pred))
print("R² Score:", r2_score(y_test, y_pred))
print("MAE:",mean_absolute_error(y_test,y_pred))

plt.scatter(df['cgpa'],df['package'])
plt.plot(X_train,lr.predict(X_train),color='black')
plt.xlabel('CGPA')
plt.ylabel('Package(in lpa)')

# y = mx + b
m = lr.coef_   #value of slope
b = lr.intercept_  #value of intercept
#package=m * cgpa + b

print("Predicted package for CGPA 8.58:", m * 8.58 + b,"lpa")
print("Predicted package for CGPA 9.5:", m * 9.5 + b,"lpa")
print("Predicted package for CGPA 6.46:", m * 6.46 + b,'lpa')
```
